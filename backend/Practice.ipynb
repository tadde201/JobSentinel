```python
import pandas as pd
import numpy as np
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
import col

spark = SparkSession.builder.appName("Practice").getOrCreate()
```

```python
transactions = [
    ("A", "2026-10-01", 100),
    ("A", "2026-10-02", 150),
    ("A", "2026-10-04", 50),
    ("B", "2026-10-01", 200),
    ("B", "2026-10-03", 300),
    ("B", "2026-10-04", 100),
]

cols = ["customer_id", "order_date", "amount"]

# Pandas
txp = pd.DataFrame(transactions, columns=cols)
txp["order_date"] = pd.to_datetime(txp["order_date"])

```

```python
print(txp)
```

```python
grouped = (
    txp.groupby("customer_id")
      .agg(
          total_amount=("amount", "sum"),
          average_amount=("amount", "mean"),
          transaction_count=("amount", "count")
      )
)

print(grouped)
```

```python
#Removed the duplicate code and kept the relevant parts.
removedup=txp.drop_duplicates(subset=["customer_id", "order_date"], keep="first")
```

```python
print(removedup)
```

```python
txp = txp.sort_values(
    ["customer_id", "order_date"]
).copy()

txp["running_total"] = (
    txp.groupby("customer_id")["amount"].cumsum()
)
print(txp)
```

```python
# Use pandas aggregation since the data is a pandas DataFrame, not a Spark DataFrame
# installing pyspark is not required for this notebook

grouped = (
    txp.groupby("customer_id")
      .agg(
          total_amount=("amount", "sum"),
          average_amount=("amount", "mean"),
          transaction_count=("amount", "count")
      )
)

print(grouped)
```

```python
# This notebook uses pandas, so keep the aggregation in pandas syntax.
grouped = (
    txp.groupby("customer_id")
      .agg(
          total_amount=("amount", "sum"),
          average_amount=("amount", "mean"),
          transaction_count=("amount", "count")
      )
)

print(grouped)
```

```python
txp = pd.DataFrame(transactions, columns=cols)



```

```python
spark_df = spark.createDataFrame(txp)
```

```python
# PySpark is not needed for this notebook because the data is already a pandas DataFrame.


grouped = (
    spark_df.groupBy("customer_id")
    .agg(
        F.sum("amount").alias("total_amount"),
        F.avg("amount").alias("average_amount"),
        F.count("amount").alias("transaction_count")
    )
)

grouped.show()
```

```python
# Final aggregation for the current pandas DataFrame

grouped = (
    txp.groupby("customer_id")
      .agg(
          total_amount=("amount", "sum"),
          average_amount=("amount", "mean"),
          transaction_count=("amount", "count")
      )
)

print(grouped)
```
